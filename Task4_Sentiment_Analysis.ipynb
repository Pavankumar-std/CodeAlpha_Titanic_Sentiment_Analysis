```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from textblob import TextBlob

# Sample customer reviews dataset
data = {
    "Review": [
        "This product is amazing and works perfectly!",
        "I really love the quality of this product.",
        "The product is okay and does its job.",
        "I am very disappointed with the product.",
        "Terrible experience, I will not buy again.",
        "Excellent quality and fast delivery!",
        "The product arrived on time.",
        "Very bad quality and poor customer service.",
        "I am happy with my purchase.",
        "Not good, the product stopped working quickly.",
        "Great value for money!",
        "The service was average."
    ]
}

df = pd.DataFrame(data)

df
```

```python
def get_sentiment(text):
    polarity = TextBlob(text).sentiment.polarity
    
    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"

df["Sentiment"] = df["Review"].apply(get_sentiment)

df
```

```python
print(df["Sentiment"].value_counts())
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Sentiment",
    order=["Positive", "Neutral", "Negative"]
)

plt.title("Customer Review Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")

plt.show()
```

```python
df["Polarity"] = df["Review"].apply(
    lambda x: TextBlob(x).sentiment.polarity
)

df[["Review", "Sentiment", "Polarity"]]
```
